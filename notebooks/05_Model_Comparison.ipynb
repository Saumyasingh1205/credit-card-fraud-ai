# Model Comparison

This notebook compares the performance of the machine learning models developed for credit card fraud detection.

The models evaluated are:

- Logistic Regression
- Random Forest
- XGBoost

The models are compared using accuracy, precision, recall, and F1-score.

```python
import pandas as pd
import matplotlib.pyplot as plt
```

## Model Performance Results

The evaluation results obtained from the individual model notebooks are consolidated into a single table for comparison.

The models were evaluated on the same test dataset using the same evaluation metrics.

```python
results = {
    "Logistic Regression": {
        "Accuracy": 99.91,
        "Precision": 84.62,
        "Recall": 57.89,
        "F1 Score": 68.75
    },
    "Random Forest": {
        "Accuracy": 99.95,
        "Precision": 97.18,
        "Recall": 72.63,
        "F1 Score": 83.13
    },
    "XGBoost": {
        "Accuracy": 99.95,
        "Precision": 94.52,
        "Recall": 72.63,
        "F1 Score": 82.14
    }
}
```

## Comparison Table

The performance metrics of Logistic Regression, Random Forest, and XGBoost are presented below.

```python
comparison_df = pd.DataFrame(results).T

comparison_df
```

## Best Model by Metric

The best-performing model for each evaluation metric is identified to understand the strengths of the different approaches.

```python
print("Best Accuracy :", comparison_df["Accuracy"].idxmax())
print("Best Precision:", comparison_df["Precision"].idxmax())
print("Best Recall   :", comparison_df["Recall"].idxmax())
print("Best F1 Score :", comparison_df["F1 Score"].idxmax())
```

## Performance Visualization

The following bar chart compares the performance of the three models across Accuracy, Precision, Recall, and F1 Score.

```python
comparison_df.plot(
    kind="bar",
    figsize=(10, 6)
)

plt.title("Model Performance Comparison")
plt.xlabel("Model")
plt.ylabel("Score (%)")
plt.xticks(rotation=0)
plt.legend(title="Metric")
plt.tight_layout()
plt.show()
```

## Recall Comparison

Recall is particularly important in credit card fraud detection because it measures the proportion of actual fraudulent transactions correctly identified by the model.

A low recall means that the model is missing fraudulent transactions.

Therefore, recall is an important metric when selecting a model for fraud detection.

```python
comparison_df["Recall"].plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title("Recall Comparison")
plt.xlabel("Model")
plt.ylabel("Recall (%)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## F1 Score Comparison

The F1 Score is the harmonic mean of Precision and Recall.

It provides a balanced measure of model performance, particularly when the dataset contains a significant class imbalance.

A higher F1 Score indicates a better balance between correctly identifying fraudulent transactions and minimizing false positive predictions.

```python
comparison_df["F1 Score"].plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title("F1 Score Comparison")
plt.xlabel("Model")
plt.ylabel("F1 Score (%)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

# Model Comparison Analysis

## Logistic Regression

Logistic Regression provides the baseline performance for the fraud detection problem.

It achieved:

- Accuracy: 99.91%
- Precision: 84.62%
- Recall: 57.89%
- F1 Score: 68.75%

Although the accuracy is high, its relatively lower recall indicates that it misses a considerable number of fraudulent transactions.

## Random Forest

Random Forest achieved:

- Accuracy: 99.95%
- Precision: 97.18%
- Recall: 72.63%
- F1 Score: 83.13%

It improved substantially over Logistic Regression in precision, recall, and F1 Score.

The higher recall indicates that Random Forest detects more fraudulent transactions while maintaining high precision.

## XGBoost

XGBoost achieved:

- Accuracy: 99.95%
- Precision: 94.52%
- Recall: 72.63%
- F1 Score: 82.14%

XGBoost also significantly outperformed Logistic Regression.

It achieved the same displayed recall as Random Forest while maintaining high precision and F1 Score.

# Overall Comparison

The comparison shows that both Random Forest and XGBoost significantly outperform the Logistic Regression baseline.

Random Forest achieved the highest Precision and F1 Score among the three models.

Random Forest and XGBoost achieved the same displayed Recall of approximately 72.63%, which is substantially higher than the 57.89% Recall achieved by Logistic Regression.

The results demonstrate that tree-based ensemble models are better suited for capturing the nonlinear relationships present in the credit card transaction data.

# Conclusion

Based on the evaluation results, Random Forest provides the best overall performance among the three models.

It achieved:

- Accuracy: 99.95%
- Precision: 97.18%
- Recall: 72.63%
- F1 Score: 83.13%

XGBoost also performed strongly, achieving:

- Accuracy: 99.95%
- Precision: 94.52%
- Recall: 72.63%
- F1 Score: 82.14%

Logistic Regression serves as a useful baseline but has considerably lower Recall and F1 Score.

Since the dataset is highly imbalanced, Accuracy alone is not sufficient for selecting the best model. Precision, Recall, and F1 Score provide a more meaningful evaluation of fraud detection performance.

Based on the current results, Random Forest is selected as the preferred model for the next stage of the project.
